## 3. Create Sample Data

We'll create sample data in the SQLite database for the dashboard to display.

```python
# Create and verify sample data
import os, sys
os.chdir('/content/dataset')
sys.path.insert(0, os.getcwd())

from recruitment_functions import create_sample_data, load_recruitment_data
print('Creating sample data...')
create_sample_data(force=True)

# Verify data was created
df = load_recruitment_data()
print(f'\nSample data created successfully. Loaded {len(df)} rows.')
print('\nColumns available:', ', '.join(df.columns))
```

## 4. Start Dashboard

Finally, we'll start the Streamlit dashboard and expose it via ngrok.

```python
# Start Streamlit and expose via ngrok
import os, subprocess, time
from pyngrok import ngrok

REPO_DIR = '/content/dataset'
PORT = 8501
os.chdir(REPO_DIR)

print('Starting ngrok tunnel...')
tunnel = ngrok.connect(PORT)
print('Public URL:', tunnel.public_url)

logfile = os.path.join(REPO_DIR, 'streamlit_demo.log')
cmd = f'nohup streamlit run dashboard.py --server.port {PORT} --server.headless true > {logfile} 2>&1 &'
print('\nStarting Streamlit...')
subprocess.call(cmd, shell=True)
time.sleep(3)

print('\nStreamlit logs:')
!tail -n 40 /content/dataset/streamlit_demo.log || true
print('\nOpen the public URL above in your browser to view the dashboard.')
```
