## Analyze key metrics

### Subtask:
Calculate key performance indicators (KPIs) relevant to a recruitment company, such as number of applications, interview stages, successful placements, time to hire, and source effectiveness.


**Reasoning**:
Interpret the placeholder columns as recruitment metrics and calculate the specified KPIs.



```python
# Interpret placeholder columns as recruitment metrics
# population: 'number of applicants'
# households: 'number of companies hiring' or 'successful placements' (depending on KPI)
# total_rooms: 'total number of interview stages'
# median_income: 'salary offered'
# median_house_value: 'placement value'

# KPI 1: Placement Rate (Interpreting households as successful placements and population as applicants)
# This assumes each household represents a successful placement and population represents applicants.
# It's a simplification due to the placeholder data.
total_applicants = df['population'].sum()
successful_placements = df['households'].sum()

# Avoid division by zero if there are no applicants
placement_rate = (successful_placements / total_applicants) * 100 if total_applicants > 0 else 0

# KPI 2: Average Interview Stages per Placement (Interpreting total_rooms as interview stages and households as placements)
# This assumes total_rooms is the sum of all interview stages across all placements (households).
# It's a simplification due to the placeholder data.
total_interview_stages = df['total_rooms'].sum()

# Avoid division by zero if there are no placements
average_interview_stages_per_placement = total_interview_stages / successful_placements if successful_placements > 0 else 0

# KPI 3: Average Placement Value (Using median_house_value as a proxy for placement value)
average_placement_value = df['median_house_value'].mean()

# Store KPIs in a dictionary
recruitment_kpis = {
    'Placement Rate (%)': placement_rate,
    'Average Interview Stages per Placement': average_interview_stages_per_placement,
    'Average Placement Value': average_placement_value
}

# Print the calculated KPIs
print("Calculated Recruitment KPIs (based on placeholder data interpretation):")
for kpi, value in recruitment_kpis.items():
    print(f"{kpi}: {value:.2f}")
```

## Add interactivity

### Subtask:
Implement interactive elements like filters, drill-downs, or tooltips to enhance user experience and allow for deeper exploration of the data.


**Reasoning**:
Convert the previously generated matplotlib bar chart into an interactive chart using plotly and add tooltips to display the KPI values.



```python
import plotly.graph_objects as go

kpi_names = list(recruitment_kpis.keys())
kpi_values = list(recruitment_kpis.values())

fig = go.Figure(data=[go.Bar(
    x=kpi_names,
    y=kpi_values,
    text=[f'{v:.2f}' for v in kpi_values], # Add text labels to bars
    textposition='auto', # Position text automatically
    marker_color=['skyblue', 'lightgreen', 'salmon'], # Match colors from matplotlib
    hoverinfo='text', # Show only text on hover
    hovertext=[f'{name}: {value:.2f}' for name, value in zip(kpi_names, kpi_values)] # Tooltip text
)])

fig.update_layout(
    title="Interactive Recruitment Key Performance Indicators (Based on Placeholder Data)",
    xaxis_title="KPI",
    yaxis_title="Value",
    xaxis=dict(tickangle=45), # Rotate labels
    bargap=0.2 # Add space between bars
)

fig.show()
```

## Query Data with SQL

To query the data with SQL, we'll first load our pandas DataFrames into a temporary SQLite database in Colab.

```python
import sqlite3
import pandas as pd # Import pandas here as well, just in case it wasn't imported earlier in the session

# Connect to an in-memory SQLite database
# This database exists only for the duration of the Colab session
conn = sqlite3.connect(':memory:')

# Load the main DataFrame (df) into a SQL table
# The table name will be 'recruitment_data'
df.to_sql('recruitment_data', conn, index=False, if_exists='replace')

# Load the KPIs dictionary into a SQL table
# Convert the dictionary to a pandas Series, then to a DataFrame for easier loading
kpi_series = pd.Series(recruitment_kpis) # Convert dictionary to Series
kpi_df = kpi_series.reset_index() # Convert Series to DataFrame and reset index
kpi_df.columns = ['KPI_Name', 'Value'] # Rename columns for clarity
kpi_df.to_sql('recruitment_kpis', conn, index=False, if_exists='replace')

print("Data loaded into temporary SQLite database tables: 'recruitment_data' and 'recruitment_kpis'")
```
